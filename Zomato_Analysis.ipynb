#Training and testing





**Define the prediction goal**

```python
target_column = "Dining_Rating"
features = ["Dining_Review_Count", "Delivery_Rating", "Pricing_for_2"]
```

 **Split the data into training and testing sets**

```python
from sklearn.model_selection import train_test_split
X_train,X_test, y_train,y_test = train_test_split(merge_data[features], merge_data[target_column], test_size=0.2, random_state=42)
```

```python
X_train.shape
```

**Initialize models**

```python
from sklearn.ensemble import RandomForestRegressor

random_forest_model = RandomForestRegressor(random_state=42)

```

**Train the models**

```python
random_forest_model.fit(X_train, y_train)
```

**Make predictions on the test set**

```python
random_forest_predictions = random_forest_model.predict(X_test)
```

```python
X_train_predict=random_forest_model.predict(X_train)
```

**Calculating mean_square_error**

```python
import sklearn.metrics

random_forest_model_mse=sklearn.metrics.mean_squared_error(y_test,random_forest_predictions )
```

**Print mean_square_error**

```python
print(f"random_forest_model_mean squared error :{random_forest_model_mse}")

```

**Printing Score of Model**

```python
print("-------------------------")
print("r2 score of random forest model is:",random_forest_model.score(X_train,y_train))
print("-------------------------")
print("r2 score of random forest model is:",random_forest_model.score(X_test,y_test))
```

**Plot the predictions against the actual values**

```python
# Plot the predictions against the actual values
plt.figure(figsize=(10, 6))
plt.scatter(y_test, random_forest_predictions, label="Random Forest", alpha=0.5)
plt.xlabel("Actual Dining Ratings")
plt.ylabel("Predicted Dining Ratings")
plt.title("Comparison of Prediction Models")
plt.legend()
plt.show()
```
